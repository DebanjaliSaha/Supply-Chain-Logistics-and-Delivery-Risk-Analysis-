```python
 # Outlier Detection using IQR method
Q1 = df['Order Profit Per Order'].quantile(0.25)
Q3 = df['Order Profit Per Order'].quantile(0.75)
IQR = Q3 - Q1

lower_boundary = Q1 - 1.5 * IQR
upper_boundary = Q3 + 1.5 * IQR

outliers = df[(df['Order Profit Per Order'] < lower_boundary) | 
              (df['Order Profit Per Order'] > upper_boundary)]

print(f'Q1: {Q1}')
print(f'Q3: {Q3}')
print(f'IQR: {IQR}')
print(f'Lower Boundary: {lower_boundary}')
print(f'Upper Boundary: {upper_boundary}')
print(f'Number of Outliers: {len(outliers)}')
print(f'Percentage of Outliers: {len(outliers)/len(df)*100:.2f}%')
```

```python
# Skewness and Kurtosis
skewness = df[['Delay Days', 'Sales', 'Order Profit Per Order']].skew()
kurtosis = df[['Delay Days', 'Sales', 'Order Profit Per Order']].kurt()

print("Skewness:")
print(skewness)
print("\nKurtosis:")
print(kurtosis)
```

```python
# Shipping Mode Segmentation 
Shipping_performance = df.groupby('Shipping Mode')['Is Late'].mean() * 100
Shipping_performance= Shipping_performance.sort_values(ascending=False)
Shipping_performance
```

```python
def classify_risk(x):
    if x >= 70:
        return 'High Risk'
    elif x >= 50:
        return 'Medium Risk'
    else:
        return 'Low Risk'
      
Shipping_segment = Shipping_performance.reset_index()
Shipping_segment['Risk Level'] = Shipping_segment['Is Late'].apply(classify_risk)
Shipping_segment
```

```python
region_performance = df.groupby('Order Region')['Is Late'].mean() * 100
region_performance = region_performance.sort_values(ascending=False)
region_performance
```

```python
region_segment = region_performance.reset_index()
region_segment['Risk Level'] = region_segment['Is Late'].apply(classify_risk)

region_segment
```

```python
category_performance = df.groupby('Category Name')['Delay Days'].mean().sort_values(ascending=False)

category_segment = category_performance.reset_index()
category_segment
```
